## Data Cleaning

```python
# Check the number of missing values in each column
df.isna().sum()
```

```python
# Remove rows that contain at least one missing value
df.dropna(inplace=True)
```

##### After checking for missing values, the `dropna()` operation was applied.Since the dataset contains no missing entries, the number of rows remained unchanged. This indicates good data quality at this stage of the analysis.


```python
# Check for duplicated rows number in the dataset
df.duplicated().sum()
```

##### All occurrences of duplicated rows were examined to understand the extent of repetition in the dataset. By looking at every instance, we can confirm if duplicates exist and determine whether they are meaningful or should be removed. In this dataset, no significant duplicated observations were found, indicating good data quality.

```python
#Check for duplicated occurences in the dataset
df[df.duplicated(keep=False)]
```

```python
# Check for duplicated rows in the dataset
df[df.duplicated()]
```

```python
# Remove duplicated rows
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

## Exploratory Data Analysis (EDA)

```python
# Count Fruit
df['fruit_name'].value_counts()
```

```python
# Count Fruit by shape
df.groupby(['shape'])['fruit_name'].value_counts()
```

```python
# Display the max of Fruit
df.groupby(['shape'])['fruit_name'].value_counts().idxmax()
```

```python
df.groupby('shape')['fruit_name'].agg(lambda x: x.value_counts().idxmax())
```

```python
df["weight (g)"].mean()
```

```python
df["size (cm)"].describe()
```

```python
# Calculate average weight of fruits for each shape
df.groupby('shape')["weight (g)"].mean()
```

```python
df.groupby('fruit_name')["avg_price (₹)"].mean()
```

## Creation two new analytical metrics

```python
# Density Index: weight per size
df['density'] = df["weight (g)"] / df["size (cm)"]
df
```

```python
df['price_per_gram'] = df["avg_price (₹)"] / df["weight (g)"]
df
```

## Data Visualization with Matplotlib

```python
df["fruit_name"].value_counts().plot(kind="bar", legend=True)
plt.title("Fruit Name Distribution")
```

```python

df.groupby('fruit_name')["avg_price (₹)"].mean().plot(kind="pie")
plt.title("Pie Chart of Fruit by average Price")
```
